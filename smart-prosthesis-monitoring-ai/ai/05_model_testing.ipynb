# 05 - Model Testing

Evaluates the trained model on the held-out test split: confusion matrix,
accuracy, precision, recall, F1 - same evaluation style as thesis Figure 3.3.4.

```python
import sys
sys.path.insert(0, "../src")
import pandas as pd
import joblib
import matplotlib.pyplot as plt
from sklearn.metrics import (
    confusion_matrix, ConfusionMatrixDisplay, accuracy_score,
    precision_score, recall_score, f1_score
)
from prediction import split_features, predict_risk, FEATURE_COLUMNS

features = pd.read_csv("../data/features.csv")
X_train, X_test, y_train, y_test = split_features(features)
model = joblib.load("../results/risk_model.joblib")

preds = predict_risk(model, X_test)
```

```python
cm = confusion_matrix(y_test, preds)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Safe (0)", "Risk (1)"])
disp.plot(cmap="Blues", values_format="d")
plt.title("Confusion Matrix - Risk Prediction (reconstruction)")
plt.savefig("../results/confusion_matrix_reconstruction.png", dpi=120)
plt.show()
```

```python
metrics = {
    "accuracy": accuracy_score(y_test, preds),
    "precision": precision_score(y_test, preds, zero_division=0),
    "recall": recall_score(y_test, preds, zero_division=0),
    "f1": f1_score(y_test, preds, zero_division=0),
}
metrics
```
