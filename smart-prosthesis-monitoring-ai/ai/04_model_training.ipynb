**Model choice note:** the thesis does not document a specific model architecture
for the predictive AI component - only its inputs, outputs, and the patterns it
should recognize. Rather than inventing an architecture and presenting it as the

solves (recognizing repeated high/medium-load routines) as the **thesis's own

```python
import sys
sys.path.insert(0, "../src")
import pandas as pd
from prediction import build_model, split_features, FEATURE_COLUMNS

features = pd.read_csv("../data/features.csv")
X_train, X_test, y_train, y_test = split_features(features)
print("Train:", X_train.shape, " Test:", X_test.shape)
```

```python
model = build_model()
model.fit(X_train, y_train)

import joblib
joblib.dump(model, "../results/risk_model.joblib")
print("Model trained and saved to ../results/risk_model.joblib")
```
