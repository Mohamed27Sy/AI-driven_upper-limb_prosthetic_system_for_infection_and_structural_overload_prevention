# 06 - Results Analysis

Summarizes model performance and interprets it the way the thesis frames its own
results: a demonstration of the methodology on a laboratory/simulation dataset,
**not** a claim of clinical validation.

```python
import sys
sys.path.insert(0, "../src")
import pandas as pd
import joblib
from sklearn.metrics import classification_report
from prediction import split_features, predict_risk

features = pd.read_csv("../data/features.csv")
X_train, X_test, y_train, y_test = split_features(features)
model = joblib.load("../results/risk_model.joblib")
preds = predict_risk(model, X_test)

print(classification_report(y_test, preds, target_names=["Safe", "Risk"]))
```

## Interpretation

The model demonstrated strong performance recognizing repeated high- and
medium-load routines on this synthetic evaluation set, consistent with the shape
of the results reported in the original thesis (Figure 3.3.4). This should **not**
be read as "the model is clinically validated" - as the thesis itself states,
testing was limited to laboratory/artificial-environment simulations, not real
patients or real implant hardware. See `../results/results_summary.md` and the
**Limitations** section of the top-level `README.md` for the full caveats.
